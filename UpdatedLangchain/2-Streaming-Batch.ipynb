## <b style="color: green;">Streaming & Batch </b> ##

#### <b style="color: orange;">Streaming</b> ####

<b>Streaming is a technique in which data is delivered and processed continuously in small chunks instead of waiting for the entire response to be generated.</b>

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")

```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
response = model.invoke("hello")
response.content[0]["text"]
```

|Below Down It will create a generator Object


```python
model.stream("write 200 words on Artificial Intelligence")
```

```python
for chunk in model.stream("Write 200 words on Artificial Intelligence"):
    print(chunk.text, end="|", flush=True)
```

#### <b style="color: orange;">Batching</b> ####

<b>Batching is a technique in which multiple inputs are processed together in a single operation instead of processing each input separately.</b>

```python
responses = model.batch(
    [
        "hello",
        "what is AI",
        "what is quantum Coputing?"
    ],
    config={
        'max_concurrency':5,
    }
)
for response in responses:
    for chunk in model.stream(response.content[0]["text"]):
        print(chunk.text, end="|", flush=True)
```
