#### <b style="color: orange;"> Model Integration with Google Gemini </b> ####

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
```

#### <b style="color: brown;">1st Using ChatGoogleGenerativeAI </b> ####

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
response = model.invoke("hello")
response
```

#### <b style="color: brown;">2nd Using init_chat_model</b> ####

```python
from langchain.chat_models import init_chat_model

model = init_chat_model("google_genai:gemini-3.5-flash-lite")
response = model.invoke("hello, what is my name")
response.content[0]["text"]
```
