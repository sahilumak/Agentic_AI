## <b style="color: green;">Structured output </b> ##

Structured output means asking an LLM to return information in a predefined format instead of ordinary, free-form text.

### <b style="color: brown;">Pydantic </b> ###

Pydantic models provides the richest features set with field validation, descriptions, and nested structures.
Pydantic is a Python library that defines data structures using classes and validates that the data matches the specified types and rules.
