```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
model
```

```python
from langchain.tools import tool

@tool
def get_price(product:str)-> int:
    """Get the Price of Specific Product"""
    return f"Price of {product} is 500rs"
model_with_tool = model.bind_tools([get_price])
```

```python
model_with_tool.invoke("what is the price of laptop")
```

```python
message = [{"role":"user","content":"What is the price of laptop, is it more than 500rs"}]
ai_msg = model_with_tool.invoke(message)
message.append(ai_msg)

for tool_call in ai_msg.tool_calls:
    print(tool_call)
    tool_result = get_price.invoke(tool_call["args"])
    # print(tool_result)
    message.append(tool_result)
    # print(messages)

final_response = model_with_tool.invoke(message)
print(final_response.text)
```
